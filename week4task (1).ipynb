```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Create sample dataset
data = {
    'Date': [
        '2026-01-01', '2026-01-02', '2026-01-03',
        '2026-01-04', '2026-01-05', '2026-01-06',
        '2026-01-07', '2026-01-08', '2026-01-09',
        '2026-01-10', '2026-01-11', '2026-01-12'
    ],
    'Sales': [
        100, 120, 110, 130, 140, 150,
        160, 155, 140, 130, 120, 110
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python
print(df.isnull().sum())
```

```python
df = df.dropna()

print(df)
```

```python
df['Sales'] = df['Sales'].fillna(df['Sales'].mean())

print(df)
```

```python
print(df.duplicated())
```

```python
print("Duplicate Count:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print(df)
```

```python
df['Date'] = pd.to_datetime(df['Date'])

print(df.dtypes)
print(df)
```

```python
plt.figure(figsize=(10,5))

plt.plot(df['Date'], df['Sales'], marker='o')

plt.title('Sales Trend Analysis')
plt.xlabel('Date')
plt.ylabel('Sales')
plt.xticks(rotation=45)

plt.grid()
plt.show()
```

```python
df['Moving_Average'] = df['Sales'].rolling(window=3).mean()

print(df)
```

```python
plt.figure(figsize=(10,5))

plt.plot(df['Date'], df['Sales'], marker='o', label='Actual Sales')

plt.plot(df['Date'], df['Moving_Average'],
         marker='s', label='Moving Average')

plt.title('Sales Moving Average')
plt.xlabel('Date')
plt.ylabel('Sales')

plt.legend()
plt.grid()
plt.xticks(rotation=45)

plt.show()
```
